```python
#a
import numpy as np

# A matrix
A = np.array([[2,1,4,1],
              [3,4,-1,-1],
              [1,-4,1,5],
              [2,-2,1,3],],float)
# v vector
v = np.array([-4,3,9,7],float)
N = len(v)

# pivoting
for m in range(N):
    pivot_row = m + np.argmax(abs(A[m:,m]))

    # exchange
    A[[m,pivot_row]] = A[[pivot_row,m]]
    v[m], v[pivot_row] = v[pivot_row], v[m]

    # make the pivot 1
    pivot = A[m,m]
    A[m,:] /= pivot
    v[m] /= pivot

    # Subtract from the lower
    for i in range(m+1,N):
        factor = A[i,m]
        A[i,:] -= factor*A[m,:]
        v[i] -= factor*v[m]

# Backsubstitution
solution = np.empty(N)

for m in range(N-1,-1,-1):
    solution[m] = v[m]

    for i in range(m+1,N):
        solution[m] -= A[m,i]*solution[i]

print(solution)
```
