```python
#a
import numpy as np

# constants
c = 2
accuracy = 1e-6
x0 = 1
error = 1.0

# calculating the first two 
x1 = 1 - np.exp(-c*x0)
x2 = 1 - np.exp(-c*x1)

# Iteration loop
while abs(x2-x1) > accuracy: 
    # Scroll to the next step
    x0 = x1
    x1 = x2
    x2 = 1 - np.exp(-c*x1) # calculating new guess

error = (x1-x2)**2 / (2*x1-x0-x2)

print("the solution found:",x2)
print("estimated error",abs(error))



```

```python
# b
import numpy as np
import pylab as plt

# c numbers
c_values = np.arange(0,3.01,0.01)

# list for plotting
x_solutions = []
accuracy = 1e-6

for c in c_values:
    x0 = 1.0
    x1 = 1 - np.exp(-c*x0)
    x2 = 1 - np.exp(-c*x1)

    # Loop until convergence for each value of c
    for _ in range(500):
        if abs(x2-x1) < accuracy:
            break
        x0 = x1
        x1 = x2
        x2 = 1 - np.exp(-c*x1)
        
    x_solutions.append(x2)



# plotting
plt.figure(figsize=(8,5))
plt.plot(c_values,x_solutions,color='blue',linewidth=2)
plt.axvline(x=1.0,color='red',linestyle='--',label='Transition Threshold (c=1)')

plt.title('changing x solution depends on c value')
plt.xlabel('c value')
plt.ylabel('x solutions')
plt.grid(True,linestyle=':',alpha=0.6)
plt.legend()
plt.show()
```
