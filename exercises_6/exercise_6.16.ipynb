```python
"""
===============================================================================
EXERCISE 6.16 (a) - LAGRANGE POINT L1 DERIVATION (THEORETICAL PROOF)
===============================================================================

1. PROBLEM DEFINITION & FRAMEWORK:
----------------------------------
Consider a system consisting of the Earth (mass M), the Moon (mass m), and a 
satellite (mass m_s) located at the L1 Lagrange point along the line connecting 
the Earth and the Moon.

- Distance from Earth's center to Moon's center = R
- Distance from Earth's center to satellite (L1 point) = r
- Distance from satellite to Moon's center = R - r (since 0 < r < R)
- Angular orbital velocity of both the Moon and the co-rotating satellite = ω


2. FORCE BALANCE AT THE L1 LAGRANGE POINT:
------------------------------------------
The satellite at distance r experiences two opposing gravitational forces:
1. Earth's gravitational force pulling it toward the Earth (inward):
       F_Earth = (G * M * m_s) / r^2

2. Moon's gravitational force pulling it toward the Moon (outward from Earth):
       F_Moon  = (G * m * m_s) / (R - r)^2

The net gravitational force directed inward toward the Earth's center is:
       F_net = F_Earth - F_Moon
       F_net = (G * M * m_s) / r^2 - (G * m * m_s) / (R - r)^2


3. CENTRIPETAL FORCE & EQUATION DERIVATION:
-------------------------------------------
For the satellite to remain stationary relative to the Earth and Moon in this 
co-rotating frame, the net inward gravitational force must supply the required 
centripetal force for circular motion of radius r at angular frequency ω:

       F_centripetal = m_s * ω^2 * r

Setting the net gravitational force equal to the centripetal force:
       (G * M * m_s) / r^2 - (G * m * m_s) / (R - r)^2 = m_s * ω^2 * r

Dividing both sides of the equation by the satellite's mass m_s yields:

       (G * M) / r^2 - (G * m) / (R - r)^2 = ω^2 * r   (PROVED)

===============================================================================
"""
```

```python
# b
import numpy as np

# constants
G = 6.674e-11
M = 5.974e24
m = 7.348e22
R = 3.844e8
w = 2.662e-6
accuracy = 1e-4

def f(r):
    return G*M/(r**2) - G*m/(R-r)**2 - w**2 * r

r1 = 3.0e8
r2 = 3.5e8

delta = 1.0

while abs(delta) > accuracy:
    f_r1 = f(r1)
    f_r2 = f(r2)
    r3 = r2 - f_r2 * (r2 - r1) / (f_r2 - f_r1)

    delta = r3 -r2

    r1 = r2
    r2 = r3

print(f"Distance to the L1 Lagrange Point r = {r2:.4e} m")
```
