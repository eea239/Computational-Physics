```python
"""
Exercise 6.5 - Part (a) Theoretical Derivation and Proof
=========================================================

1. FUNDAMENTAL PRINCIPLES AND DEFINITIONS
-----------------------------------------
According to Kirchhoff's Current Law (KCL), the sum of all currents leaving a node is equal to zero (sum(I) = 0).

- Input Voltage     : V_+ = x_+ * e^(i*omega*t)
- Node Voltages     : V_1 = x_1 * e^(i*omega*t), V_2 = x_2 * e^(i*omega*t), V_3 = x_3 * e^(i*omega*t)
- Resistor Current  : I_R = (V_a - V_b) / R
- Capacitor Current : I_C = C * d(V_a - V_b) / dt

Using the differentiation rule:
d/dt [ (x_a - x_b) * e^(i*omega*t) ] = i * omega * (x_a - x_b) * e^(i*omega*t)


2. DERIVATION OF NODE EQUATIONS
-------------------------------

📍 NODE 1:
(V_1 - V_+) / R_1  +  V_1 / R_4  +  C_1 * d(V_1 - V_2) / dt = 0

Substituting the expressions for V and their derivatives:
(x_1 - x_+) * e^(i*omega*t) / R_1 + x_1 * e^(i*omega*t) / R_4 + i * omega * C_1 * (x_1 - x_2) * e^(i*omega*t) = 0

Dividing both sides by e^(i*omega*t) to simplify:
(x_1 - x_+) / R_1 + x_1 / R_4 + i * omega * C_1 * (x_1 - x_2) = 0

Factoring out x_1 and x_2, and moving x_+ to the right-hand side [EQUATION 1]:
(1/R_1 + 1/R_4 + i*omega*C_1)*x_1 - i*omega*C_1*x_2 = x_+ / R_1


📍 NODE 2:
(V_2 - V_+) / R_2  +  V_2 / R_5  +  C_1 * d(V_2 - V_1) / dt  +  C_2 * d(V_2 - V_3) / dt = 0

Substituting V expressions and simplifying by e^(i*omega*t):
(x_2 - x_+) / R_2 + x_2 / R_5 + i * omega * C_1 * (x_2 - x_1) + i * omega * C_2 * (x_2 - x_3) = 0

Factoring out x_1, x_2, and x_3 [EQUATION 2]:
-i*omega*C_1*x_1 + (1/R_2 + 1/R_5 + i*omega*C_1 + i*omega*C_2)*x_2 - i*omega*C_2*x_3 = x_+ / R_2


📍 NODE 3:
(V_3 - V_+) / R_3  +  V_3 / R_6  +  C_2 * d(V_3 - V_2) / dt = 0

Substituting V expressions and simplifying by e^(i*omega*t):
(x_3 - x_+) / R_3 + x_3 / R_6 + i * omega * C_2 * (x_3 - x_2) = 0

Factoring out x_2 and x_3 [EQUATION 3]:
-i*omega*C_2*x_2 + (1/R_3 + 1/R_6 + i*omega*C_2)*x_3 = x_+ / R_3


3. CONCLUSION
-------------
Combining these three derived equations yields the 3x3 linear system presented in the textbook.
"""
```

```python
# b
import numpy as np

# parameters
R1 = R3 = R5 = 1000
R2 = R4 = R6 = 2000
C1 = 1e-6
C2 = 0.5e-6
X_plus = 3
omega = 1000

constant1_1 = (1/R1 + 1/R4 + 1j*omega*C1)
constant1_2 = (1j*omega*C1)
constant2_1 = (1j*omega*C1)
constant2_2 = (1/R2 + 1/R5 + 1j*omega*C1 + 1j*omega*C2)
constant2_3 = (1j*omega*C2)
constant3_1 = (1j*omega*C2)
constant3_2 = (1/R3 + 1/R6 + 1j*omega*C2)

right_1 = X_plus / R1
right_2 = X_plus / R2
right_3 = X_plus / R3

# A matrix
A = np.array([[constant1_1,-constant1_2,0],
              [-constant2_1,constant2_2,-constant2_3],
              [0,-constant3_1,constant3_2]
])

# v vector
v = np.array([right_1,right_2,right_3],float)

# solution
solution = np.linalg.solve(A,v)


# amplitude
x1,x2,x3 = solution

z1 = np.abs(x1)
z2 = np.abs(x2)
z3 = np.abs(x3)

print('Amplitude:',z1,z2,z3)

# phase
phase1 = np.angle(x1,deg=True)
phase2 = np.angle(x2,deg=True)
phase3 = np.angle(x3,deg=True)

print('Phase angle:',phase1,phase2,phase3)


```
