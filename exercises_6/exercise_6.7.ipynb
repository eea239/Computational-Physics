```python
"""
EXERCISE 6.7 (a) - THEORETICAL DERIVATION & MATRIX STRUCTURE
============================================================

1. THEORETICAL PROOF (Kirchhoff's Current Law & Ohm's Law):
-----------------------------------------------------------
At any junction (node), the sum of net currents must be zero: sum(I) = 0
Ohm's Law: I = (V_node - V_neighbor) / R

Multiplying each node equation by R to clear the denominator:

* Node V1 (Neighbors: V+, V2, V3):
  (V1 - V+) + (V1 - V2) + (V1 - V3) = 0
  =>  3*V1 - V2 - V3 = V+

* Node V2 (Neighbors: V+, V1, V3, V4):
  (V2 - V+) + (V2 - V1) + (V2 - V3) + (V2 - V4) = 0
  => -V1 + 4*V2 - V3 - V4 = V+

* Internal Nodes Vi (2 < i < N-1) (Neighbors: Vi-2, Vi-1, Vi+1, Vi+2):
  (Vi - Vi-2) + (Vi - Vi-1) + (Vi - Vi+1) + (Vi - Vi+2) = 0
  => -Vi-2 - Vi-1 + 4*Vi - Vi+1 - Vi+2 = 0

* Node VN-1 (Neighbors: VN-3, VN-2, VN, Ground=0V):
  (VN-1 - VN-3) + (VN-1 - VN-2) + (VN-1 - VN) + (VN-1 - 0) = 0
  => -VN-3 - VN-2 + 4*VN-1 - VN = 0

* Node VN (Neighbors: VN-2, VN-1, Ground=0V):
  (VN - VN-2) + (VN - VN-1) + (VN - 0) = 0
  => -VN-2 - VN-1 + 3*VN = 0


2. DEFINITION OF MATRIX A AND VECTOR w (Av = w Form):
----------------------------------------------------
- Matrix A (N x N pentadiagonal / 5-banded matrix):
    * Main diagonal       (offset 0)  : [3, 4, 4, ..., 4, 4, 3]
    * 1st off-diagonals   (offset ±1) : [-1, -1, ..., -1]
    * 2nd off-diagonals   (offset ±2) : [-1, -1, ..., -1]

- Vector w (N-dimensional right-hand side vector):
    * w[0] = V+
    * w[1] = V+
    * w[2:] = 0
"""

import numpy as np

# function
def constructing_of_A_and_w(N,V_plus=5.0):
    A = (
        4 * np.eye(N)
        - np.eye(N,k=1)
        - np.eye(N,k=-1)
        - np.eye(N,k=2)
        - np.eye(N,k=-2)

    )
    # Apply boundary conditions
    A[0,0] = 3.0
    A[-1,-1] = 3.0

    # w vector
    w = np.zeros(N,dtype=float)
    w[0] = V_plus
    w[1] = V_plus

    solution = np.linalg.solve(A,w)
    return solution

    
```

```python
# b
N = 6
A, w = constructing_of_A_and_w(N, V_plus=5.0)

# solution
solution = np.linalg.solve(A,w)
print(solution)



```

```python
# c
import scipy.linalg as sc
import numpy as np

N = 10000
V_plus = 5.0

# A matrix
ab = np.zeros((5, N))

ab[0, 2:] = -1.0  # Upper 2
ab[1, 1:] = -1.0  # Upper 1
ab[2, :] = 4.0  # Diagonal
ab[2, 0] = 3.0  # Border condition
ab[2, -1] = 3.0  # Border condition
ab[3, :-1] = -1.0  # Lower 1
ab[4, :-2] = -1.0  # Lower 2

# w vector
w = np.zeros(N, dtype=float)
w[0] = V_plus
w[1] = V_plus

solution = sc.solve_banded((2, 2), ab, w)

print(solution)
```
