```python
"""
===============================================================================
EXERCISE 6.13 (a) - WIEN'S DISPLACEMENT LAW (THEORETICAL PROOF)
===============================================================================

1. PROBLEM DEFINITION & PLANCK'S RADIATION LAW:
-----------------------------------------------
According to Planck's radiation law, the radiation intensity per unit area and 
per unit wavelength λ is given by:
    I(λ) = (2 * π * h * c^2 * λ^-5) / (e^(h*c / (λ*k_B*T)) - 1)

To find the wavelength λ at which the emitted radiation is strongest (maximum point),
we must set the derivative of I(λ) with respect to λ to zero: dI/dλ = 0.


2. DIFFERENTIATION & ALGEBRAIC REARRANGEMENT (dI/dλ = 0):
----------------------------------------------------------
For simplicity, let us define constants:
    C = 2 * π * h * c^2
    A = (h * c) / (k_B * T)

Thus, I(λ) can be expressed as:
    I(λ) = C * λ^-5 * (e^(A/λ) - 1)^-1

Applying the product rule (u * v)' = u' * v + u * v':
    u = λ^-5                    => u' = -5 * λ^-6
    v = (e^(A/λ) - 1)^-1        => v' = -1 * (e^(A/λ) - 1)^-2 * e^(A/λ) * (-A * λ^-2)
                                      = (A / λ^2) * e^(A/λ) * (e^(A/λ) - 1)^-2

Setting the derivative to zero:
    dI/dλ = C * [ -5 * λ^-6 * (e^(A/λ) - 1)^-1  +  λ^-5 * (A / λ^2) * e^(A/λ) * (e^(A/λ) - 1)^-2 ] = 0

Factoring out [ C * λ^-6 * (e^(A/λ) - 1)^-2 ] and dividing both sides by it gives:
    -5 * (e^(A/λ) - 1) + λ * (A / λ^2) * e^(A/λ) = 0
    -5 * e^(A/λ) + 5 + (A / λ) * e^(A/λ) = 0

Dividing the entire equation by e^(A/λ) (or multiplying by e^(-A/λ)):
    -5 + 5 * e^(-A/λ) + (A / λ) = 0

Rearranging terms:
    5 * e^(-A/λ) + (A / λ) - 5 = 0

Substituting back A = (h * c) / (k_B * T) yields the required equation:
    5 * e^(-h*c / (λ*k_B*T)) + (h*c / (λ*k_B*T)) - 5 = 0  (PROVED)


3. SUBSTITUTION & DERIVATION OF WIEN'S DISPLACEMENT LAW:
--------------------------------------------------------
Make the substitution x = (h * c) / (λ * k_B * T) into the equation above:
    5 * e^(-x) + x - 5 = 0  (Nonlinear equation in x)

Solving for λ from the definition of x:
    x = (h * c) / (λ * k_B * T)
    => λ = (h * c) / (k_B * x * T)

By defining the Wien displacement constant b = (h * c) / (k_B * x):
    λ = b / T

Thus, it is analytically proven that the wavelength of maximum radiation obeys 
Wien's displacement law, with b defined in terms of the solution x.
===============================================================================
"""
```

```python
# b

import numpy as np

def binary_search(f,x1,x2,accuracy=1e-6,max_iter=100):
    if f(x1) * f(x2) >= 0:
        print('error:f(x1) and f(x2) are not opposite sign! Method Does not work')  
        return None
    step = 0

    while abs(x1-x2) > accuracy and step < max_iter:
        step += 1

        # calculating middle point
        x_mid = (x1+x2)/2.0
        # If the exact root is found, exit the loop
        if f(x_mid) == 0.0:
            break
        # Narrow the range by half using sign checking.
        if f(x_mid) * f(x1) > 0:
            x1 = x_mid  # Update left border if f(x_mid) and f(x1) have the same sign
        else:
            x2 = x_mid  # Update right border if opposite sign

    # Final estimate: Midpoint of the remaining interval
    root = (x1 + x2) / 2.0
    print(f"the root was found in {step}.")
    return root


# solution

def f(x):
    return 5* np.exp(-x) + x - 5 

x1 = 1.0
x2 = 6.0

root = binary_search(f,x1,x2)
print(root)
```
