```python
"""
===============================================================================
EXERCISE 6.17 (a) - NONLINEAR CIRCUITS (THEORETICAL DERIVATION)
===============================================================================

1. CIRCUIT TOPOLOGY & DIODE MODEL:
----------------------------------
Consider a modified Wheatstone bridge circuit with four resistors (R1, R2, R3, R4) 
and a diode connected between the nodes with voltages V1 and V2.

- Top rail potential = V+
- Bottom rail potential = 0 V (Ground)
- Diode points from node V1 to node V2 (forward voltage V = V1 - V2)
- Current flowing through diode from V1 to V2 is given by:
      I_D = I_0 * [ e^((V1 - V2) / V_T) - 1 ]


2. APPLICATION OF KIRCHHOFF'S CURRENT LAW (KCL) AT NODE V2:
------------------------------------------------------------
Kirchhoff's Current Law states that the sum of all currents leaving a junction 
must equal zero:
      Σ I_leaving = 0

For junction V2, let us sum all outgoing currents:

1. Current leaving V2 toward top supply rail V+ through resistor R3:
       I_R3 = (V2 - V+) / R3

2. Current leaving V2 toward ground (0 V) through resistor R4:
       I_R4 = (V2 - 0) / R4 = V2 / R4

3. Current leaving V2 toward node V1 through the diode:
       Since diode current I_D naturally flows from V1 into V2, the current 
       leaving V2 towards V1 is the negative of I_D:
       I_diode_out = -I_D = -I_0 * [ e^((V1 - V2) / V_T) - 1 ]


3. FINAL NONLINEAR EQUATION FOR NODE V2:
----------------------------------------
Summing all three outgoing currents at node V2 to zero:

       I_R3 + I_R4 + I_diode_out = 0

       (V2 - V+) / R3  +  V2 / R4  -  I_0 * [ e^((V1 - V2) / V_T) - 1 ] = 0   (DERIVED)

===============================================================================
"""
```

```python
# b
import numpy as np

# constants
V_plus = 5.0
R1 = 1000
I0 = 3e-9
R2 = 4000
V_T = 0.05
R3 = 3000
R4 = 2000
accuracy = 1e-4

def F(v):
    v1, v2 = v[0], v[1]

    f1 = (v1 - V_plus) / (R1) + v1/R2 + I0 * (np.exp((v1-v2)/V_T)-1)
    f2 = (v2 - V_plus) / (R3) + v2 / R4 - I0 * (np.exp((v1-v2) / V_T)-1)

    return np.array([f1,f2])

def J(v):
    v1, v2 = v[0] , v[1]
    exp_term = (I0 / V_T) * np.exp((v1 - v2) / V_T)
    df1_dv1 , df1_dv2 = (1 / R1) + (1 / R2) + exp_term, -exp_term
    df2_dv1, df2_dv2 = -exp_term, (1 / R3) + (1 / R4) + exp_term
    return np.array([[df1_dv1, df1_dv2],
                     [df2_dv1, df2_dv2]])

v_guess = np.array([1.0,1.0])
delta = np.array([1.0,1.0])

while np.linalg.norm(delta) > accuracy:
    delta = np.linalg.solve(J(v_guess), F(v_guess))
    v_guess = v_guess - delta

# found roots
v1_result = v_guess[0]
v2_result = v_guess[1]

print('v1 =',v1_result,'v2 =',v2_result)

```
