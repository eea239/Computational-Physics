```python
# a
import numpy as np
import pylab as plt

# x-axis
x = np.linspace(0,1,100)

# function
y = 924*(x**6) - 2772*(x**5) + 3150*(x**4) - 1680*(x**3) + 420*(x**2) - 42*x +1

# plotting
plt.figure(figsize=(8,5))
plt.plot(x,y,color='blue',linewidth=1.5)
plt.xlabel('x')
plt.ylabel('function values')
plt.grid(True,alpha=0.6)
plt.show()
```

```python
# b
import numpy as np

# accuracy
accuracy = 1e-10

# guessing of root
guess = [0.03, 0.17, 0.38, 0.62, 0.83, 0.97]

# list
real_root = []

for x in guess:
    delta = 1.0  
    
    while abs(delta) > accuracy:  
        
        y_derivate = 5544*(x**5) - 13860*(x**4) + 12600*(x**3) - 5040*(x**2) + 840*x - 42
        y = 924*(x**6) - 2772*(x**5) + 3150*(x**4) - 1680*(x**3) + 420*(x**2) - 42*x + 1
        
        delta = y / y_derivate
        x = x - delta
    
    real_root.append(x)  

print("calculated roots:")
for i, r in enumerate(real_root, 1):
    print(f"root {i}: {r:.10f}")







```
